This notebook ingests images (binaryFile), metadata (CSV), and training images (binaryFile). Then schedule it in an existing Job.

```python
# Claims images ingestion

images_df = (
  spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "binaryFile")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/main/default/claims_vol/autoloader_metadata/images_schema"
    )
    .load("/Volumes/main/default/claims_vol/claims/images")
)

(images_df.writeStream
   .format("delta")
   .option(
       "checkpointLocation",
       "/Volumes/main/default/claims_vol/autoloader_metadata/images_chk"
   )
   .trigger(availableNow=True)
   .toTable("workspace.bronze.claim_images"))
```

```python
# Claims metadata (CSV)

metadata_df = (
  spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "csv")
    .option("header", "true")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/main/default/claims_vol/autoloader_metadata/metadata_schema"
    )
    .load("/Volumes/main/default/claims_vol/claims/metadata")
)

(metadata_df.writeStream
   .format("delta")
   .option(
       "checkpointLocation",
       "/Volumes/main/default/claims_vol/autoloader_metadata/metadata_chk"
   )
   .trigger(availableNow=True)
   .toTable("workspace.bronze.claim_metadata"))
```

```python
# Training images

training_df = (
  spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "binaryFile")
    .option(
        "cloudFiles.schemaLocation",
        "/Volumes/main/default/training_imgs_vol/autoloader_metadata/schema"
    )
    .load("/Volumes/main/default/training_imgs_vol/training_imgs")
)

(training_df.writeStream
   .format("delta")
   .option(
       "checkpointLocation",
       "/Volumes/main/default/training_imgs_vol/autoloader_metadata/chk"
   )
   .trigger(availableNow=True)
   .toTable("workspace.bronze.training_images"))
```
